```python

import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist, squareform
import matplotlib.pyplot as plt
# 1. Dataset sederhana
data = np.array([
 [1.0, 1.0], # a
 [1.5, 1.5], # b
 [5.0, 5.0], # c
 [3.0, 4.0], # d
 [4.0, 4.0], # e
 [3.0, 3.5] # f
])

# 2. Hitung Distance Matrix (Euclidean Distance)
distance_matrix = pdist(data, metric='euclidean')
distance_df = pd.DataFrame(squareform(distance_matrix),
 columns=['a', 'b', 'c', 'd', 'e', 'f'],
 index=['a', 'b', 'c', 'd', 'e', 'f'])

# 3. Tampilkan Distance Matrix
print("=== Distance Matrix (Euclidean) ===")
print(distance_df)

# 4. Hierarchical Clustering (Single Linkage)
linked = linkage(data, method='single')

# 5. Plot Dendrogram
plt.figure(figsize=(12, 8))
plt.title("Hierarchical Clustering Dendrogram (Single Linkage)")
dendrogram(linked, labels=['a', 'b', 'c', 'd', 'e', 'f'], leaf_rotation=90)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

Kode tersebut digunakan untuk melakukan Hierarchical Clustering dengan metode Single Linkage pada dataset sederhana. Prosesnya dimulai dengan menyiapkan library yang diperlukan, membuat data, menghitung jarak antar data, kemudian menampilkan hasil pengelompokan dalam bentuk dendrogram.

Pertama, program mengimpor beberapa library. numpy digunakan untuk membuat data, pandas digunakan untuk menampilkan data dalam bentuk tabel, scipy digunakan untuk menghitung jarak dan melakukan clustering, sedangkan matplotlib digunakan untuk membuat grafik.

Setelah itu, program menghitung jarak antar data menggunakan Euclidean Distance dengan pdist(). Hasil jarak tersebut kemudian diubah menjadi bentuk tabel menggunakan squareform() dan DataFrame().
```python
import numpy as np
import pandas as pd
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist, squareform
import matplotlib.pyplot as plt

# 1. Dataset sederhana
data = np.array([
    [1.0, 1.0],  # a
    [1.5, 1.5],  # b
    [5.0, 5.0],  # c
    [3.0, 4.0],  # d
    [4.0, 4.0],  # e
    [3.0, 3.5]   # f
])

# 2. Hitung Distance Matrix (Euclidean Distance)
distance_matrix = pdist(data, metric='euclidean')

distance_df = pd.DataFrame(
    squareform(distance_matrix),
    columns=['a', 'b', 'c', 'd', 'e', 'f'],
    index=['a', 'b', 'c', 'd', 'e', 'f']
)

# 3. Tampilkan Distance Matrix
print("=== Distance Matrix (Euclidean) ===")
print(distance_df)

# 4. Hierarchical Clustering
linked_single = linkage(data, method='single')
linked_complete = linkage(data, method='complete')
linked_average = linkage(data, method='average')

# 5. Plot Dendrogram Single Linkage
plt.figure(figsize=(10, 6))
plt.title("Hierarchical Clustering Dendrogram (Single Linkage)")
dendrogram(
    linked_single,
    labels=['a', 'b', 'c', 'd', 'e', 'f'],
    leaf_rotation=90
)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()

# 6. Plot Dendrogram Complete Linkage
plt.figure(figsize=(10, 6))
plt.title("Hierarchical Clustering Dendrogram (Complete Linkage)")
dendrogram(
    linked_complete,
    labels=['a', 'b', 'c', 'd', 'e', 'f'],
    leaf_rotation=90
)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()

# 7. Plot Dendrogram Average Linkage
plt.figure(figsize=(10, 6))
plt.title("Hierarchical Clustering Dendrogram (Average Linkage)")
dendrogram(
    linked_average,
    labels=['a', 'b', 'c', 'd', 'e', 'f'],
    leaf_rotation=90
)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.grid(True)
plt.show()
```

Kode tersebut digunakan untuk melakukan Hierarchical Clustering pada dataset sederhana yang terdiri dari enam data, yaitu a, b, c, d, e, dan f. Pertama, library yang diperlukan seperti numpy, pandas, scipy, dan matplotlib dipanggil untuk mengolah data, menghitung jarak, melakukan clustering, dan menampilkan grafik.

Selanjutnya, data dihitung jaraknya menggunakan Euclidean Distance dengan pdist(). Hasil jarak tersebut kemudian diubah menjadi bentuk Distance Matrix menggunakan squareform() dan ditampilkan dalam bentuk tabel agar jarak antar setiap data dapat dilihat dengan lebih mudah.
Setelah itu, dilakukan Hierarchical Clustering menggunakan tiga metode, yaitu:
single menggabungkan cluster berdasarkan jarak terdekat, complete berdasarkan jarak terjauh, sedangkan average berdasarkan rata-rata jarak antar data dalam cluster.

Kemudian, masing-masing hasil clustering ditampilkan dalam bentuk dendrogram. Dendrogram digunakan untuk melihat bagaimana data-data tersebut bergabung menjadi beberapa kelompok berdasarkan jaraknya.


Kesimpulan:
Perbandingan: Single linkage menggabungkan cluster berdasarkan jarak terdekat, complete linkage berdasarkan jarak terjauh, sedangkan average linkage berdasarkan rata-rata jarak antar data dalam cluster. Oleh karena itu, bentuk dan ketinggian dendrogram yang dihasilkan ketiga metode berbeda. Single linkage cenderung menggabungkan data pada jarak yang lebih rendah, sedangkan complete linkage membutuhkan jarak yang lebih besar. Average linkage menghasilkan pengelompokan yang berada di antara kedua metode tersebut.
