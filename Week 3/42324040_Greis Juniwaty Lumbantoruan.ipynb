```python
import pandas as pd
df = pd.read_csv('Salary_Data.csv')
print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
import matplotlib.pyplot as plt
plt.scatter(X, y, color = "red")
plt.xlabel('Years Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#bagi dataset jadi train dan test 80:20
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, 
random_state = 0)
```

```python
# Fitting Simple Linear Regression to the Training set
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
a = regressor.intercept_
b = regressor.coef_[0]
print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
# memprediksi hasil pada data uji (Test set) menggunakan model regresi 
# yang tela
y_pred = regressor.predict(X_test)
y_pred
```

```python
# Menghitung nilai R-Square, yang mengukur seberapa baik model regresi linear
# Semakin mendekati 1, semakin baik model
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
score
```

```python
# Visualizing the Training set results
plt.scatter(X_train, y_train, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
```

```python
# Visualizing the Test set results
plt.scatter(X_test, y_test, color = 'red')
plt.plot(X_train, regressor.predict(X_train), color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#TUGAS NO 1
import pandas as pd
df = pd.read_csv('Student_Scores.csv')
print(df.head())
```

```python
#Menentukan X dan y
X = df.iloc[:, :-1].values
y = df.iloc[:, 1].values
```

```python
#Visualisasi data
import matplotlib.pyplot as plt

plt.scatter(X, y, color='red')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.title('Study Hours vs Score')
plt.show()
```

```python
#Membagi dataset menjadi training dan testing
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0
)
```

```python
#Membuat dan melatih model
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train, y_train)
```

```python
#Menentukan intercept dan slope
a = regressor.intercept_
b = regressor.coef_[0]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b:", b)
print(f"ŷ = {a:.2f} + {b:.2f}x")
```

```python
#Melakukan prediksi pada data testing
y_pred = regressor.predict(X_test)

print(y_pred)
```

```python
#Menghitung nilai R²
from sklearn.metrics import r2_score

score = r2_score(y_test, y_pred)

print("R-Square:", score)
```

```python
#visualisasi hasil regresi
#Training set
plt.scatter(X_train, y_train, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')

plt.title('Score vs Study Hours (Training Set)')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.show()
```

```python
#Test set
plt.scatter(X_test, y_test, color='red')
plt.plot(X_train, regressor.predict(X_train), color='blue')

plt.title('Score vs Study Hours (Test Set)')
plt.xlabel('Study Hours')
plt.ylabel('Score')
plt.show()
```

```python
#TUGAS NO 2
import pandas as pd
df2 = pd.read_csv('Marketing_Data.csv')
print(df2.head())
```

```python
#Menentukan predictor (X) dan target (y)
X2 = df2.iloc[:, :-1]
y2 = df2.iloc[:, 3]
```

```python
#Membagi data menjadi training dan testing
from sklearn.model_selection import train_test_split

X_train2, X_test2, y_train2, y_test2 = train_test_split(
    X2, y2, test_size=0.2, random_state=0
)
```

```python
#Membuat dan melatih model
from sklearn.linear_model import LinearRegression

regressor = LinearRegression()
regressor.fit(X_train2, y_train2)
```

```python
#Melakukan prediksi
y_pred2 = regressor.predict(X_test2)

print("Predicted Sales:")
print(y_pred2)
```

```python
#Menentukan intercept dan koefisien
a = regressor.intercept_
b1 = regressor.coef_[0]
b2 = regressor.coef_[1]
b3 = regressor.coef_[2]

print("Estimated model intercept, a:", a)
print("Estimated model slope, b1:", b1)
print("Estimated model slope, b2:", b2)
print("Estimated model slope, b3:", b3)

print(f"ŷ = {a:.2f} + {b1:.2f}x1 + {b2:.2f}x2 + {b3:.2f}x3")
```

```python
#Menghitung R-Square
from sklearn.metrics import r2_score

score2 = r2_score(y_test2, y_pred2)

print("R-Square:", score2)
```
