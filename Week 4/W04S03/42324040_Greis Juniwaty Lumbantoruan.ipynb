```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
# Load dataset Iris dari scikit-learn
iris = load_iris()
```

Kode ini digunakan untuk mengimpor library yang diperlukan dan mengambil dataset Iris. numpy digunakan untuk mengolah data dalam bentuk array, sedangkan pandas digunakan untuk mengelola data dalam bentuk tabel/DataFrame. load_iris digunakan untuk mengambil dataset Iris yang sudah tersedia di scikit-learn, kemudian iris = load_iris() menyimpan dataset tersebut ke dalam variabel iris agar dapat digunakan pada tahap selanjutnya.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Kode ini digunakan untuk mengubah dataset Iris menjadi DataFrame dan memisahkan data fitur dengan target. pd.DataFrame() membuat tabel dari data Iris dengan nama kolom sesuai fitur Iris. Kolom Species ditambahkan untuk menyimpan nama jenis bunga. Kemudian X berisi fitur-fitur bunga tanpa kolom Species, sedangkan y berisi target atau jenis bunga Iris. print(X) digunakan untuk menampilkan data fitur yang akan digunakan dalam proses klasifikasi.

```python
y
```

y digunakan untuk menentukan target atau label yang akan diprediksi. Variabel y mengambil kolom Species dari DataFrame df, yang berisi jenis bunga Iris seperti setosa, versicolor, dan virginica. Jadi, y merupakan data target yang akan digunakan oleh model klasifikasi.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Kode ini digunakan untuk mengubah label species yang awalnya berupa teks menjadi angka agar dapat diproses oleh algoritma Naive Bayes. LabelEncoder() melakukan proses encoding pada y, sehingga label seperti setosa, versicolor, dan virginica diubah menjadi nilai numerik 0, 1, dan 2. Hasil encoding kemudian disimpan kembali ke y dan ditampilkan.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Kode ini digunakan untuk membagi data menjadi data training dan data testing. test_size=0.2 berarti 20% data digunakan untuk testing dan 80% untuk training, sedangkan random_state=0 digunakan agar pembagian data tetap sama setiap kali kode dijalankan. X_train dan y_train digunakan untuk melatih model, sedangkan X_test dan y_test digunakan untuk menguji hasil prediksi model.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Kode ini digunakan untuk membuat dan melatih model Gaussian Naive Bayes. GaussianNB() membuat model Naive Bayes yang sesuai untuk data numerik kontinu seperti dataset Iris. Kemudian model.fit(X_train, y_train) melatih model menggunakan data training X_train dan label y_train, lalu model yang sudah dilatih disimpan dalam variabel NB_model.

```python
result = NB_model.predict(X_test)
result
```

Kode ini digunakan untuk memprediksi kelas/species dari data testing menggunakan model NB_model yang sudah dilatih sebelumnya. NB_model.predict(X_test) menghasilkan prediksi untuk setiap data pada X_test, kemudian hasilnya disimpan dalam variabel result dan ditampilkan.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Kode ini digunakan untuk menghitung tingkat akurasi model berdasarkan hasil prediksi result dan data sebenarnya y_test. accuracy_score() membandingkan keduanya, kemudian dikalikan 100 agar hasil akurasi ditampilkan dalam bentuk persentase. Pada modul, hasil akurasi data testing adalah sekitar 96,67%.

Kode ini digunakan untuk memprediksi species dari data Iris baru menggunakan model NB_model yang sudah dilatih. input_data berisi 4 nilai fitur bunga, kemudian NB_model.predict() menentukan kelasnya. Hasil prediksi yang awalnya berupa angka diubah kembali menjadi nama species menggunakan lab.inverse_transform(). Untuk input [5.1, 3.5, 1.4, 0.2], hasil prediksinya adalah setosa.
